```python
import pandas as pd
import sys
 
ORIGINAL_DATASET_PATH = "dataset.csv" 
SEPARATOR = "||"
 
def generate_solution(filled_path: str, output_path: str = "solution.csv"):
    original = pd.read_csv(ORIGINAL_DATASET_PATH)
    filled   = pd.read_csv(filled_path)
 
    feature_cols = [c for c in original.columns if c != "datetime"]
 
    rows = []
    for col in feature_cols:
        was_missing = original[col].isna()
 
        for idx in original.index[was_missing]:
            dt  = original.loc[idx, "datetime"]
            uid = f"{dt}{SEPARATOR}{col}"
            val = filled.loc[idx, col]
            rows.append({"id": uid, "value": val})
 
    solution = pd.DataFrame(rows, columns=["id", "value"])
    solution = solution.sort_values("id").reset_index(drop=True)
    solution.to_csv(output_path, index=False)
    print(f"✅ Solution saved → {output_path}  ({len(solution)} rows)")
 
# generate_solution("filled_dataset.csv", "submission.csv")
# generate_solution("filled_dataset_knn.csv", "submission_knn.csv")

# generate_solution(
#     "filled_dataset_cubic.csv",
#     "submission_cubic.csv"
# )

# generate_solution(
#     "filled_dataset_rf.csv",
#     "submission_rf.csv"
# )

# generate_solution(
#     "filled_dataset_surface_v2.csv",
#     "submission_surface_v2.csv"
# )


# generate_solution(
#     "filled_dataset_surface_v3.csv",
#     "submission_surface_v3.csv"
# )

# generate_solution(
#     "filled_dataset_surface_v4.csv",
#     "submission_surface_v4.csv"
# )

# generate_solution(
#     "filled_dataset_surface_v5_80.csv",
#     "submission_surface_v5_80.csv"
# )

# generate_solution(
#     "filled_dataset_surface_v6.csv",
#     "submission_surface_v6.csv"
# )

# generate_solution(
#     "filled_dataset_surface_v7.csv",
#     "submission_surface_v7.csv"
# )

# generate_solution(
#     "filled_dataset_xgb.csv",
#     "submission_xgb.csv"
# )

# generate_solution(
#     "filled_dataset_surface_v8.csv",
#     "submission_surface_v8.csv"
# )

# generate_solution(
#     "filled_dataset_time_v1.csv",
#     "submission_time_v1.csv"
# )

# generate_solution(
#     "filled_dataset_time_v2.csv",
#     "submission_time_v2.csv"
# )

# generate_solution(
#     "filled_dataset_blend_90_10.csv",
#     "submission_blend_90_10.csv"
# )

# generate_solution(
#     "filled_dataset_blend_95_5.csv",
#     "submission_blend_95_5.csv"
# )

# generate_solution(
#     "filled_dataset_hybrid_95_5.csv",
#     "submission_hybrid_95_5.csv"
# )

# generate_solution(
#     "filled_dataset_hybrid_xgb_99_1.csv",
#     "submission_hybrid_xgb_99_1.csv"
# )

# generate_solution(
#     "filled_dataset_hybrid_70_30.csv",
#     "submission_hybrid_70_30.csv"
# )

# generate_solution(
#     "filled_dataset_hybrid_50_50.csv",
#     "submission_hybrid_50_50.csv"
# )

# generate_solution(
#     "filled_dataset_hybrid_75_25.csv",
#     "submission_hybrid_75_25.csv"
# )

# generate_solution(
#     "filled_dataset_time_strike_v2.csv",
#     "submission_pure_strike.csv"
# )

# generate_solution(
#     "filled_dataset_pure_strike_smooth.csv",
#     "submission_pure_strike_smooth.csv"
# )

# generate_solution(
#     "filled_dataset_pure_strike_smooth_v2.csv",
#     "submission_pure_strike_smooth_v2.csv"
# )

# generate_solution(
#     "filled_dataset_strike_xgb_90_10.csv",
#     "submission_strike_xgb_90_10.csv"
# )

generate_solution(
    "filled_dataset_double_smooth_final.csv",
    "submission_double_smooth_final.csv"
)
```
